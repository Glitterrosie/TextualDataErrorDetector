```python
import pandas as pd

# data = pd.read_csv('../datasets/imdb_subset1_group1_w_errors.csv')
# data = pd.read_csv('../datasets/weather_subset1_group1_w_errors.csv')
# data = pd.read_csv('../datasets/medical_subset1_group1_w_errors.csv')

# data = pd.read_csv('../../TabularRedundancyPolluter/datasets/imdb_subset2_group2.csv')
# data = pd.read_csv('../../TabularRedundancyPolluter/datasets/weather_subset2_group2.csv')
data = pd.read_csv('../../TabularRedundancyPolluter/datasets/medical_subset2_group2.csv')

uniques = {}
for column in data.columns:
    unique_values = data[column].unique()
    uniques[column] = unique_values

max_unique = max(len(values) for values in uniques.values())
unique_frame = pd.DataFrame(columns=data.columns, index=range(max_unique))
for column, values in uniques.items():
    unique_frame[column] = pd.Series(values)

display(unique_frame)

```

### Get list of unique values for column

```python

column_name = 'discharge_disposition_desc'
if not column_name in unique_frame.columns:
    raise ValueError(f"Column '{column_name}' not found in the dataset.")

unique_column_values = unique_frame[column_name].dropna().unique().tolist()
print(f"{column_name.upper()}_VALUES = [")
for value in unique_column_values:
    print(f"    '{value.strip()}',")
print("]")
```
