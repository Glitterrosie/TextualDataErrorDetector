```python
import pandas as pd

# dataset_name = 'imdb'
# dataset_name = 'weather'
dataset_name = 'medical'

data = pd.read_csv(f"../datasets/{dataset_name}_subset1_group1_w_errors.csv")
labels = pd.read_csv(f"../datasets/{dataset_name}_subset1_group1_error_mappings.csv")

no_label_mask = labels == 0
label_mask = labels != 0

label_values = data.copy()


label_values[no_label_mask] = None
# Convert values to strings and prepend the label value
for col in data.columns:
    for idx in data.index:
        if label_mask.loc[idx, col]:
            label_int = labels.loc[idx, col]
            original_value = data.loc[idx, col]
            label_values.loc[idx, col] = f"{label_int}: {original_value}"



display(label_values)
```
